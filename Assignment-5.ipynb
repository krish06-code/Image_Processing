```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("scenery.jpg", 0)

hist, bins = np.histogram(img.flatten(), 256, [0, 256])
cdf = hist.cumsum()
cdf_normalized = (cdf - cdf.min()) * 255 / (cdf.max() - cdf.min())
equalized = np.interp(img.flatten(), range(256), cdf_normalized).reshape(img.shape).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap='gray')
plt.title("Equalized Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("scenery.jpg", 0)
ref = cv2.imread("reference.jpg", 0)
if ref is None:
    ref = np.clip(img * 1.5, 0, 255).astype(np.uint8)

hist_src, _ = np.histogram(img.flatten(), 256, [0, 256])
hist_ref, _ = np.histogram(ref.flatten(), 256, [0, 256])

cdf_src = hist_src.cumsum() / hist_src.sum()
cdf_ref = hist_ref.cumsum() / hist_ref.sum()

lut = np.zeros(256, dtype=np.uint8)
for i in range(256):
    lut[i] = np.argmin(np.abs(cdf_src[i] - cdf_ref))

matched = lut[img]

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(ref, cmap='gray')
plt.title("Reference Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(matched, cmap='gray')
plt.title("Matched Image")
plt.axis("off")

plt.show()
```
