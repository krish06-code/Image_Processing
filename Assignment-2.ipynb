```python
import cv2
import numpy as np

img = cv2.imread("/content/scenery.jpg")

B, G, R = cv2.split(img)

gray = ((R + G + B) / 3).astype(np.uint8)

cv2.imshow("Original", img)
cv2.imshow("Grayscale", gray)

```

```python
import cv2
import numpy as np

img = cv2.imread("/content/scenery.jpg")

B, G, R = cv2.split(img)

wr = float(input("Enter weight of R: "))
wg = float(input("Enter weight of G: "))
wb = float(input("Enter weight of B: "))

if wr + wg + wb != 1:
    print("Sum of weights must be 1")
else:
    gray = (wr * R + wg * G + wb * B).astype(np.uint8)

    cv2.imshow("Original", img)
    cv2.imshow("Grayscale", gray)

```
