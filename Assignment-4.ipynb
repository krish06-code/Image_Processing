```python
import cv2
import numpy as np
import math
import matplotlib.pyplot as plt

img = cv2.imread("/content/scenery.jpg", 0)
L=256
c=L/math.log(L+1)

result = np.zeros(img.shape, dtype=np.float32)


for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        result[i][j] = c*math.log(1 + float(img[i][j]))


result = (result / result.max()) * 255
result = np.uint8(result)


plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result, cmap='gray')
plt.title("Log Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/scenery.jpg", 0)

gamma = float(input("Enter gamma value: "))

c = 1

result = np.zeros(img.shape, dtype=np.float32)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        result[i][j] = c * (float(img[i][j]) ** gamma)


result = (result / result.max()) * 255
result = np.uint8(result)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result, cmap="gray")
plt.title("Power Law Transformation")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np

img = cv2.imread("/content/scenery.jpg", 0)

r1 = int(input("Enter input lower value: "))
r2 = int(input("Enter input upper value: "))
s1 = int(input("Enter output lower value: "))
s2 = int(input("Enter output upper value: "))

output = img.copy()

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        if r1 <= img[i][j] <= r2:
            output[i][j] = ((img[i][j] - r1) * (s2 - s1) / (r2 - r1)) + s1

cv2.imshow("Original", img)
cv2.imshow("Output", output)

 
```
