```python
import cv2
import numpy as np

img = cv2.imread("/content/scenery.jpg", 0)

threshold = np.mean(img)

binary = np.where(img >= threshold, 255, 0).astype(np.uint8)

cv2.imshow("Grayscale", img)
cv2.imshow("Binary", binary)

 
```
