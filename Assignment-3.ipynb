```python
import cv2

img = cv2.imread("/content/scenery.jpg", cv2.IMREAD_GRAYSCALE)

width = int(input("Enter border width: "))
color = int(input("Enter border color (0-255): "))

output = cv2.copyMakeBorder(
    img,
    width, width, width, width,
    cv2.BORDER_CONSTANT,
    value=color
)

cv2.imshow("Original Image", img)
cv2.imshow("Image with Border", output)

 
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/scenery.jpg', 0)

gray_comp = 255 - img

_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
binary_comp = 255 - binary

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_comp, cmap='gray')
plt.title("Grayscale Complement")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(binary, cmap='gray')
plt.title("Binary")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(binary_comp, cmap='gray')
plt.title("Binary Complement")
plt.axis("off")

plt.show()
```
