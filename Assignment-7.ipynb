```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("scenery.jpg", 0)

laplacian = cv2.Laplacian(img, cv2.CV_64F)
sharpened = np.clip(img - laplacian, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sharpened, cmap='gray')
plt.title("Laplacian Sharpened")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("scenery.jpg", 0)

sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobel_mag = np.sqrt(sobelx**2 + sobely**2)
sobel_sharp = np.clip(img + 0.5 * sobel_mag, 0, 255).astype(np.uint8)

canny = cv2.Canny(img, 100, 200)
canny_sharp = np.clip(img + 0.5 * canny, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(sobel_mag, cmap='gray')
plt.title("Sobel Gradient")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(sobel_sharp, cmap='gray')
plt.title("Sobel Sharpened")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(canny, cmap='gray')
plt.title("Canny Edges")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(canny_sharp, cmap='gray')
plt.title("Canny Sharpened")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("scenery.jpg", 0)

blurred = cv2.GaussianBlur(img, (5, 5), 3)
mask = img.astype(np.float32) - blurred.astype(np.float32)
sharpened = np.clip(img + 1.0 * mask, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(blurred, cmap='gray')
plt.title("Gaussian Blurred (5x5, sigma=3)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharpened, cmap='gray')
plt.title("Unsharp Masking Sharpened")
plt.axis("off")

plt.show()
```
