```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("scenery.jpg", 0)

k = int(input("Enter filter size (e.g. 3, 5): "))
sigma = float(input("Enter sigma for Gaussian filter: "))
pad_method = input("Enter padding method (zero / replicate): ")

pad = k // 2
if pad_method == "zero":
    padded = cv2.copyMakeBorder(img, pad, pad, pad, pad, cv2.BORDER_CONSTANT, value=0)
else:
    padded = cv2.copyMakeBorder(img, pad, pad, pad, pad, cv2.BORDER_REPLICATE)

kernel_avg = np.ones((k, k), dtype=np.float32) / (k * k)

x = np.arange(-pad, pad + 1)
y = np.arange(-pad, pad + 1)
X, Y = np.meshgrid(x, y)
kernel_weight = np.maximum(np.abs(X), np.abs(Y)).astype(np.float32)
kernel_weight = kernel_weight / np.sum(kernel_weight)

kernel_gauss = np.exp(-(X**2 + Y**2) / (2 * sigma**2))
kernel_gauss = kernel_gauss / np.sum(kernel_gauss)

def apply_filter(padded_img, kernel):
    h, w = img.shape
    output = np.zeros((h, w), dtype=np.float32)
    for i in range(h):
        for j in range(w):
            output[i, j] = np.sum(padded_img[i:i+k, j:j+k] * kernel)
    return np.clip(output, 0, 255).astype(np.uint8)

out_avg = apply_filter(padded, kernel_avg)
out_weight = apply_filter(padded, kernel_weight)
out_gauss = apply_filter(padded, kernel_gauss)

plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(out_avg, cmap='gray')
plt.title("Averaging Filter")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(out_weight, cmap='gray')
plt.title("Weighted Filter")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(out_gauss, cmap='gray')
plt.title("Gaussian Filter")
plt.axis("off")

plt.show()
```
